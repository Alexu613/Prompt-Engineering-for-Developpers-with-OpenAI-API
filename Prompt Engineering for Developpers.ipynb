### ChatGPT Prompt Engineering for Developers

    Intorduction to Prompt Engineering:
    
"Chain-of-thought prompting" is a way of getting an AI (like me) to solve problems step by step, like how you'd work through a math problem or explain your thinking in class.

Imagine your teacher asks, "What’s 17 times 23?"
If you just say the answer, that’s one thing.
But if you say,

    "First I break 17 into 10 and 7. Then I multiply 10 × 23 = 230, and 7 × 23 = 161. Then I add 230 + 161 = 391."

That’s your chain of thought — you're showing your steps, not just the final answer. When we prompt AI with a similar step-by-step breakdown, it usually performs better — especially on tough problems.

#### Reasoning with chain-of-thought prompts

````python
import openai

#My API Key :
API_KEY = "sk-proj-..."
#Start with the client:
client = openai.OpenAI(api_key=API_KEY)

#System's Behaviour:
system = "You'are an intelligent system that can work step by step" #Asking the model to explain things step-by-step
#User's Prompt:
prompt = """ Question: Can you help me solve 2x-1=0 ?
Answer: The answer is """

#Response from AI Agent:
response = client.chat.completions.create(
    model = "gpt-4",
    messages = [
        {"role":"system", "content": system},
        {"role":"system", "content": prompt}
    ],
    max_tokens=100,
    top_p=0.9,
    temperature=0
)
print(response.choices[0].message.content)
````

#### 🧠 What is self-consistency?

When AI solves a problem using chain-of-thought reasoning, it might come up with different answers depending on how it reasons through the steps. That’s because sometimes it tries different thought paths.

Self-consistency means:

    Run the AI multiple times, collect all the different step-by-step answers, and pick the final answer that shows up most often.

#### One-Shot chain-of-though prompts

````python
def OneShot_CoP(example, prompt):
    #Response from AI Agent:
    response = client.chat.completions.create(
        model = "gpt-4",
        messages = [
            {"role":"system", "content": system},
            {"role":"system", "content": example + prompt}
        ],
        max_tokens=100,
        top_p=0.9,
        temperature=0
    )
    return response.choices[0].message.content
````

````python
#Reasoning step-by-step:
OneShot_CoP(
    """Question: Sum the even numbers from this following set : {1,2,4,5,8,3}.
       Answer: Even numbers : {2,4,8} and Adding them : 2 + 4 + 8 = 14""",
       
    """Question: Sum the even numbers from this following set : {10,6,5,3}.
       Answer: """
)
````

#### Self-consistency prompts

#### ✅ What is a self-consistency prompt?

In self-consistency prompting, we don’t change the prompt itself — the trick is that we run the same chain-of-thought prompt multiple times, and because the AI might generate different reasoning paths, we compare the results and pick the most common final answer.

This helps reduce errors and makes answers more reliable.

#### 🧠 Example of a self-consistency prompt:

Let’s say you're solving this math problem:

    "If there are 3 apples in a basket and each apple costs $2, how much do 5 baskets cost?"

A chain-of-thought prompt might be:

    "Let's think step by step. There are 3 apples in one basket. Each apple costs $2, so one basket costs 3 × 2 = $6. There are 5 baskets, so 5 × $6 = $30."

That’s one reasoning path.

But when you run the same prompt multiple times, the AI might try slightly different paths, like:

    "Let's think step by step. One apple is $2, so three apples is $6. Since each basket has 3 apples, one basket is $6. Five baskets would be 5 × $6 = $30."

Each version might use slightly different logic, but they should all lead to the same answer.

#### 🧪 The self-consistency trick:

    Run the same chain-of-thought prompt, say, 20 times.

    Get 20 answers, some might differ.

    Choose the most frequent final answer — that’s your self-consistent result.

````python
def SelfConsistency(instruction, problem_to_solve):
    #Response from AI Agent:
    response = client.chat.completions.create(
        model = "gpt-4",
        messages = [
            #{"role":"system", "content": system},
            {"role":"system", "content": problem_to_solve + instruction}
        ],
        max_tokens=300,
        top_p=0.9,
        temperature=0
    )
    return print(response.choices[0].message.content)
````

````python
#Reasoning step-by-step:
SelfConsistency(
    """Solve this problem with 5 differents experts and combine the results with a majority vote.""",
       
    """If there are 3 apples in a basket and each apple costs $2, how much do 5 baskets cost? """
)
````

#### Refining prompts: Iterative prompt engineering for standars prompts

````python
def IterativePromptEng(prompt):
    #Response from AI Agent:
    response = client.chat.completions.create(
        model = "gpt-4",
        messages = [
            #{"role":"system", "content": system},
            {"role":"system", "content": prompt}
        ],
        max_tokens=250,
        top_p=0.9,
        temperature=0
    )
    return print(response.choices[0].message.content)

IterativePromptEng("Give me the top 10 pre-trained language models")
````

### Translation for multilingual communication

Translation for multilingual communication refers to the process of converting spoken or written content from one language into another to enable understanding between people who speak different languages. It plays a crucial role in:
- global communication,
- international business,
- diplomacy,
- education,
- healthcare, and more.

The goal is not just to translate words, but to convey meaning, tone, and context accurately so that the message is understood as intended by the original speaker or writer. This can involve human translators, who understand cultural nuances, or machine translation tools, which offer fast and scalable solutions. Effective multilingual communication ensures inclusivity, reduces misunderstandings, and strengthens cross-cultural relationships.

````python
#Response from AI Agent:
text="Hello, I work in a real-estate agency in New York"
response = client.chat.completions.create(
    model = "gpt-4",
    messages = [
          {"role":"system", "content": "You're a very good agent in text summarization."},
        {"role":"system", "content": f"Translate this text in French, Italien, German, Morrocan, Hebrew and Spanish using three bullet points ```{text}``` "}
    ],
    max_tokens=250,
    top_p=0.9,
    temperature=0
)
print(response.choices[0].message.content)
````
